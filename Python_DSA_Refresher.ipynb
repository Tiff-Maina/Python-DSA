### Question 1: Days to Seconds Conversion
Write a program that asks the user for a number of days. The program then prints out the number of seconds in the amount of days given.


```python
# Ask user to enter the number of days.
days = int(input("Enter number of days: "))

# Calculate the number of seconds in the given number of days.
seconds = days * 24 * 60 * 60

# Display the result
print(f"The number of seconds in {days} days is {seconds} seconds.")
```

### Question 2: Volume of a Sphere

```python
# 2. Write a program that asks a user to input the radius then the program calculates the volume of a sphere (the formula for the volume is (4/3)πr3). Use the exponential operator in python to compute (r3).

import math
# Ask user to enter the radius of the sphere.
radius = float(input("Enter the radius of the sphere: "))

# The exponential operator (**) is used to calculate r^3.
volume = (4/3) * math.pi * (radius ** 3)

# Display the volume.
print(f"The volume of the sphere with radius {radius} is {volume} cubic units.")
```

### Question 3: Area and Perimeter of a Square

```python
# 3. Using functions, write a program to compute the area and perimeter of a square. The program should ask the user to enter a number corresponding to the side length of the square and display the area and perimeter of the square.

# This function calculates the area of the square.
def compute_area(side_length):
    return side_length ** 2

# This function calculates the perimeter of the square.
def compute_perimeter(side_length):
    return 4 * side_length

# Ask user to enter the side length of the square.
side_length = float(input("Enter the side length of the square: "))

# Compute the area and perimeter of the square.
area = compute_area(side_length)
perimeter = compute_perimeter(side_length)

# Display the results.
print(f"The area of the square with side length {side_length} is {area} square units.")
print(f"The perimeter of the square with side length {side_length} is {perimeter} units.")
```

### Question 4: Case Checker

```python
# 4. Write a program using functions that determines whether a character input by a user is uppercase or lower case.

# Define a function that checks whether a character is uppercase or lowercase
def check_character_case(char):
    if char.isupper():
        return "uppercase"
    elif char.islower():
        return "lowercase"
    else:
        return "neither uppercase nor lowercase (it may be a number or symbol)"


# Keep asking the user until they enter exactly one character
while True:
    user_char = input("Enter a single character: ")

    # Check that the user entered exactly one character
    if len(user_char) == 1:
        break
    else:
        print("Error: Please enter exactly one character.")


# Call the function to determine the character's case
result = check_character_case(user_char)

# Display the result
print(f"The character '{user_char}' is {result}.")
```

### Question 5: Pseudocode to Python 

```python
# 5. The following is pseudocode for a program being designed. Write the Python program equivalent of the same.
        # BEGIN
        # SET x TO 0, y TO 20
        # REPEAT
        # SUBTRACT 4 FROM y
        # ADD 2/y TO x
        # UNTIL
        # y IS LESS THAN 6
        # DISPLAY x
        # END

# Set the initial values of x and y
x = 0
y = 20

# Repeat the instructions until y is less than 6
while True:
    # Subtract 4 from y
    y = y - 4

    # Add 2/y to x
    x = x + (2 / y)

    # Stop the loop when y is less than 6
    if y < 6:
        break

# Display the final value of x
print(f"The final value of x is: {x}")
```

### Question 6: Average of Numbers in an Array

```python
# 6. Write a Python program that does the following:
    # Uses a loop for a user to continually input 5 values to populate an array.
    # Calculates and displays the average of the values input into the array.

# Initialize an empty list to store the values
values = []

# Use a loop to input 5 values
for i in range(5):
    value = float(input(f"Enter value {i + 1}: "))
    values.append(value)

# Calculate the average
average = sum(values) / len(values)

# Display the entered numbers and their average
print("Values entered:", values)
print(f"The average of the entered values is: {average}")
```
